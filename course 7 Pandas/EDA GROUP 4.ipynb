```python
# imports
import pandas as pd
import numpy as np
```

```python
# Data Cleaning
df = pd.read_csv('adult.csv')
df.head(20)
```

```python
# df.replace('?', pd.NA, inplace=True)

# Data cleaning for question marks (?)
df = df.map(lambda x: x.strip() if isinstance(x, str) else x)
df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)

# Data cleaning for question marks (?)
df = df.map(lambda x: x.strip() if isinstance(x, str) else x)
df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)

# df.info()
# df.head(40)
```

```python
# Column cleaning
df.columns = df.columns.str.lower().str.replace('-', '_')

# Change datatype of the column to appropraite data types
categorical_cols = ['workclass', 'education', 'marital_status', 'occupation',
                    'relationship', 'race', 'gender', 'native_country', 'income']
for col in categorical_cols:
    df[col] = df[col].astype('string')

df['age'] = df['age'].astype(int)
df['hours_per_week'] = df['hours_per_week'].astype(int)


df.info()
```

```python
# 1. How many men and women (sex feature) are represented in this dataset?
df['gender'].value_counts()
```

```python
# 2. What is the average age (age feature) of women?
age_df = df[df['gender'] == 'Female']
age_df['age'].mean()
```

```python
# 3. What is the percentage of German citizens (native-country feature)?
de_df = df[df['native_country']=='Germany']
len_ger = len(de_df)
len_tot = len(df)
result = round((len_ger / len_tot) * 100, 3)

print("The percentage is", result, "%")
```

```python
# 5. Is it true that people who earn more than 50K have at least a high school
# education? (education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, 
# Masters or Doctorate feature)

df.groupby(['education', 'income'])['income'].agg(['count'])



# Define list of "high education" levels
high_education = ['Bachelors', 'Prof-school', 'Assoc-acdm', 'Assoc-voc', 'Masters', 'Doctorate']

# Filter people who earn more than 50K
high_earners = df[df['income'] == '>50K']

# Find high earners who do NOT have high education
not_high_edu = high_earners[~high_earners['education'].isin(high_education)]

# Check how many such people exist
not_high_edu_count = len(not_high_edu)
total_high_earners = len(high_earners)

print("Number of people not having atleast Higer education" ,not_high_edu_count, "\nTotal Number having >50K:", total_high_earners)

print(round((not_high_edu_count / total_high_earners) * 100,2), "% of those not having higher education")
```

```python
# 6. Display age statistics for each race (race feature) and each gender (sex feature).
# Use groupby() and describe(). Find the maximum age of men 
# of Amer-Indian-Eskimo race.
df.head()

df.groupby(['race', 'gender'])['age'].describe()

race = df[(df['race'] == 'Amer-Indian-Eskimo') & (df['gender'] == 'Male')]['age'].max()
race
```

```python
# 7. Among whom is the proportion of those who earn a lot (>50K) greater: married or
# single men (marital-status feature)? Consider as married those who have
# a marital-status starting with Married (Married-civ-spouse, Married-spouse-absent 
# or Married-AF-spouse), the rest are considered bachelors.

# married = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']
men = df[(df['gender'] == 'Male') & (df['income'] == '>50K')]
married_men = men[men['marital_status'].str.contains('Married')]
single_men = men[~men['marital_status'].str.contains('Married')]

m_ratio = len(married_men)
s_ratio = len(single_men)

print(m_ratio, s_ratio)




# Filter only male entries
men_df = df[df['gender'] == 'Male'].copy()

# Create a new column to label men as married or single
men_df['is_married'] = men_df['marital_status'].str.startswith('Married')

# Calculate proportions of >50K earners in each group
married_men = men_df[men_df['is_married']]
single_men = men_df[~men_df['is_married']]

married_high_income_ratio = (married_men['income'] == '>50K').mean()
single_high_income_ratio = (single_men['income'] == '>50K').mean()

married_high_income_ratio, single_high_income_ratio

# len()
```

```python
# 8. What is the maximum number of hours a person works per week
# (hours-per-week feature)? How many people work such a number of hours, and
# what is the percentage of those who earn a lot (>50K) among them?

max_hour = df['hours_per_week'].max()

num_of_people = len(df[df['hours_per_week'] == max_hour])

earn_alot = df[df['income'] == '>50K']
print("The percentage is", round(len(earn_alot) / len(df) * 100, 2), "%")
```
